## Функции
Функция объявляется с помощью `def`. У аргументов могут быть значения по умолчанию, а результат возвращается через `return`:

```python
def power(x, p=2):
    """Возводит x в степень p."""
    return x ** p

print(power(3))        # p по умолчанию равно 2
print(power(2, p=10))

# функции — тоже объекты: их можно передавать в другие функции
def apply_to_list(func, values):
    return [func(v) for v in values]

print(apply_to_list(power, [1, 2, 3]))
print(apply_to_list(lambda v: v + 100, [1, 2, 3]))  # lambda — короткая безымянная функция
```

## Генераторы и списковые включения
Некоторые объекты не хранят все элементы в памяти, а вычисляют их по одному по мере перебора. Например, `range(5)` не создаёт список из пяти чисел — элементы появляются во время итерации.
Перебирать элементы в цикле `for` можно у практически всех контейнеров, а списковые включения (list comprehension) позволяют коротко создавать новые контейнеры:

```python
a = range(5)
print(a)
for i in a:
  print (i)
```

```python
b = [i for i in a]
c = [i**2 for i in a if i%2==0]
d = {i:f"Квадратом числа {i} является {i**2}" for i in c}
print("b: ", b)
print("c: ", c)
print("d: ", d)
```

Генератор — это «ленивый» объект: он хранит только текущее состояние и правило получения следующего элемента. Создать его можно функцией с `yield` или выражением в круглых скобках:

```python
def squares(n):
    for i in range(n):
        yield i ** 2

gen = squares(5)
print(gen)                 # значения ещё не вычислены
print(next(gen), next(gen))
print(list(gen))           # оставшиеся элементы

print(sum(i ** 2 for i in range(5)))  # генераторное выражение
```

### `loc` vs `iloc`

В pandas есть два способа индексирования строк:

*  по значениям столбцов и индекса (в нашем случае id) - для этого используется data.loc
*  по позиционному индексу - для этого используется data.iloc

Обратите внимание, что индексный столбец начинается с 1, поэтому позиционный индекс 0 будет соответствовать значению индексного столбца 1, позиционный 1 - значению индексного столбца 2, и так далее:

```python
print(data.index)
print('------')
print("data.iloc[0]:")
print(data.iloc[0])
print('------')
print("data.loc[1]:")
print(data.loc[1])
```

Также обратите внимание, что при индексировании с помощью `.loc` включаются оба конца среза:

```python
data.loc[2:3]
```

в то время как при использовании `.iloc` конец исключается:

```python
data.iloc[1:2]
```

### Pandas + numpy

Scikit-learn — это библиотека для построения простых конвейеров (pipeline) машинного обучения.

Она объединяет множество моделей под общим интерфейсом:

- Создание модели: `model = sklearn.whatever.ModelNameHere(parameters_if_any)`
- Обучение: `model.fit(X, y)`
- Предсказание: `model.predict(X_test)`

В ней также есть инструменты для извлечения признаков, оценки качества, кросс-валидации и многого другого.

# Метод k ближайших соседей (K-nearest neighbors)

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split


x = data[float_cols].fillna(0).copy()
y = data["failure"]

model = KNeighborsClassifier(n_neighbors=5)

# split the data into train(90%) and test(10%)
X_train, X_test, y_train, y_test = train_test_split( x, y, test_size=0.1, random_state=42)


# fit the model
model.fit(X_train,  y_train)

# make the prediction
test_predictions = model.predict(X_test)
print("Test accuracy:", accuracy_score(y_test, test_predictions))
```

Много это или мало? Сравним с простейшим «классификатором», который всегда предсказывает самый частый класс (`failure = 0`):

```python
from sklearn.dummy import DummyClassifier

baseline = DummyClassifier(strategy="most_frequent")
baseline.fit(X_train, y_train)
print("Baseline accuracy:", accuracy_score(y_test, baseline.predict(X_test)))
```

KNN оказался **хуже константы**! Всегда сравнивайте модель с простым базовым решением (baseline).

Одна из причин: KNN считает расстояния между объектами, поэтому признаки с большим разбросом значений (например, `loading`) «перевешивают» остальные. Перед KNN признаки нужно масштабировать. Объединим масштабирование и модель в цепочку (pipeline):

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

model = make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=5))
model.fit(X_train, y_train)
print("Test accuracy (with scaling):", accuracy_score(y_test, model.predict(X_test)))
```

Масштабирование — обязательный шаг для KNN, но здесь оно не помогло: константу мы так и не обогнали. Дело в том, что классы несбалансированы (≈79 % образцов не выходят из строя), а по этим признакам отказ предсказать трудно. Accuracy в такой ситуации — плохая метрика; о метриках качества поговорим на следующем занятии.

# Линейная регрессия

## y = kx + b

Начнём с одномерного случая, где истинная зависимость имеет вид:
$$y=k\cdot x+b+\text{noise}$$

```python
def linear_function(x):
  return 0.33 * x + 8.3

def gen_dataset(N, func, lims=(-1., 1.), noise_lvl=0.2):
  x = np.random.uniform(*lims, size=N)
  y = func(x) + noise_lvl * np.random.normal(size=x.shape)
  return x[:,None], y

np.random.seed(42)
X, y = gen_dataset(50, linear_function)
x = np.linspace(-1, 1, 101)
plt.plot(x, linear_function(x))
plt.scatter(X, y);
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(X, y)

x = np.linspace(-1, 1, 101)
plt.plot(x, linear_function(x), label='true function')
plt.scatter(X, y);
plt.plot(x, model.predict(x[:,None]), label='prediction')
plt.legend()

print(model.coef_, model.intercept_)
```

А что если случай более сложный?

```python
def true_function(x):
  return np.sin(3 * x + 0.8) + np.sin(1. / (x + 1.23))
```

```python
model = LinearRegression()
X, y = gen_dataset(25, true_function)

model.fit(X, y)

x = np.linspace(-1, 1, 101)
plt.plot(x, true_function(x), label='true function')
plt.scatter(X, y);
plt.plot(x, model.predict(x[:,None]), label='prediction')
plt.legend();
```

На самом деле даже довольно сложные задачи можно решать простыми алгоритмами, если хорошо понимать саму задачу!
В данном случае поможет конструирование новых признаков (фичей) — например, степеней исходного признака $x$.
Для удобства цепочку обработки данных мы будем помещать в pipeline.


$$X'=
\begin{pmatrix}
x_1 & (x_1)^2 & \ldots & (x_1)^k \\
x_2 & (x_2)^2 & \ldots & (x_2)^k \\
\vdots & \vdots & \ddots & \vdots \\
x_N & (x_N)^2 & \ldots & (x_N)^k
\end{pmatrix},
$$

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline

poly_expand = PolynomialFeatures(3)
poly_expand.fit_transform(np.arange(5)[:,None])
```

```python
# Первый параметр — степень полинома. Попробуйте разные значения: 1, 3, 5, 15, 50
poly_expand = PolynomialFeatures(5, include_bias=False)
linear_model = LinearRegression()
model = make_pipeline(
    poly_expand, linear_model
)

model.fit(X, y)

x = np.linspace(-1, 1, 101)
plt.plot(x, true_function(x), label='true function')
plt.scatter(X, y);
plt.plot(x, model.predict(x[:,None]), label='prediction')
plt.ylim(y.min() - 0.5, y.max() + 0.5)
plt.legend();
```

# Рассмотрим более сложный датасет - Boston Housing

Описание переменных:

- **CRIM** — уровень преступности на душу населения по городу
- **ZN** — доля жилой земли под участки площадью более 25 000 кв. футов
- **INDUS** — доля площадей, занятых нерозничным бизнесом, по городу
- **CHAS** — бинарная переменная «река Чарльз» (= 1, если участок граничит с рекой; 0 иначе)
- **NOX** — концентрация оксидов азота (частей на 10 млн)
- **RM** — среднее число комнат в жилище
- **AGE** — доля домов, занимаемых владельцами и построенных до 1940 г.
- **DIS** — взвешенное расстояние до пяти деловых центров Бостона
- **RAD** — индекс доступности радиальных магистралей
- **TAX** — ставка налога на недвижимость на каждые \$10 000
- **PTRATIO** — соотношение числа учеников и учителей по городу
- **B** — $1000(B_k - 0.63)^2$, где $B_k$ — доля чернокожего населения по городу
- **LSTAT** — доля населения с низким социальным статусом, %
- **MEDV** — медианная стоимость домов, занимаемых владельцами, в тысячах \$

```python
raw_df = pd.read_csv('BostonHousing.csv')
print(raw_df.head())

y = raw_df['medv']
X = raw_df.drop('medv', axis=1)
```

Мы хотим предсказать MEDV - цену жилья, исходя из остальных показателей. Оценим зависимость целевой переменной от фичей:

```python
plt.figure(figsize=(15, 15))

grid_size = int(np.ceil(X.shape[1]**0.5))

for i, (name, x) in enumerate(X.items(), 1):
  plt.subplot(grid_size, grid_size, i)
  plt.scatter(x, y)
  plt.xlabel(name)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import mean_squared_error
```

Выберем 3 наиболее удачных признака и построим модель на их основе.

```python
old_columns = X.columns
print(old_columns)
X.columns = [column.upper() for column in X.columns]
```

```python
columns = ["CRIM", "RM", "LSTAT"]

X_subset = X[columns].copy()
X_subset /= X_subset.max()
```

```python
model = make_pipeline(
    PolynomialFeatures(9, include_bias=False), # Can you calculate how many features this will result in? :)
    LinearRegression()
)

model.fit(X_subset, y)
print('mse = ', mean_squared_error(y, model.predict(X_subset)))

plt.figure(figsize=(15, 4))
for i, c in enumerate(columns, 1):
  plt.subplot(1, len(columns), i)
  plt.scatter(X[c], y, label='data')
  plt.scatter(X[c], model.predict(X_subset), label='prediction')
  plt.legend()
```

Насколько это хорошо? На вид здорово, а что на самом деле?

# Разбиение на обучающую и тестовую выборки (train-test split)

```python
X_train, X_test, y_train, y_test = train_test_split(X_subset, y, test_size=50, random_state=39)

model = make_pipeline(
    PolynomialFeatures(9, include_bias=False),
    LinearRegression()
)

model.fit(X_train, y_train)

print('train mse = ', mean_squared_error(y_train, model.predict(X_train)))
print('test mse = ', mean_squared_error(y_test, model.predict(X_test)))
```

Что?! Ошибка на тестовой выборке намного больше, чем на обучающей.

```python
plt.figure(figsize=(15, 4))
for i, c in enumerate(columns, 1):
  plt.subplot(1, len(columns), i)
  plt.scatter(X_test[c], y_test, label='data')
  plt.scatter(X_test[c], model.predict(X_test), label='prediction')
  plt.legend()
```
